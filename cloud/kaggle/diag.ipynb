# Diagnostic: what can this Kaggle session reach?

```python
import subprocess

for cmd in (
    ["cat", "/etc/resolv.conf"],
    ["getent", "hosts", "pypi.org"],
    ["getent", "hosts", "cdn.jsdelivr.net"],
    ["getent", "hosts", "huggingface.co"],
    ["getent", "hosts", "kaggle.com"],
    ["nvidia-smi", "--query-gpu=name,memory.total",
     "--format=csv,noheader"],
    ["ls", "-la", "/kaggle/input"],
    ["env"],
):
    print("$ " + " ".join(cmd))
    r = subprocess.run(cmd, capture_output=True, text=True)
    print((r.stdout or r.stderr)[:800])
    print("---")
```
